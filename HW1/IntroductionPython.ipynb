```python
energy_kJ = []
for number in energy_list_kcal:
    kJ = number * 4.184
    energy_kJ.append(kJ)

print(energy_kJ)
```

```python
#Evaluate negative items
energy_negative = []
for number in energy_kJ:
    if number < 0 or number == 0:
        energy_negative.append(number)

print(energy_negative)
```
